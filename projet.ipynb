PARTIE I


# Étude d'une matrice stochastique



On considère le vecteur :

$$
v_t =
\begin{pmatrix}
x_t \\
y_t \\
z_t
\end{pmatrix},
\qquad
x_t+y_t+z_t=1,
\qquad
x_t,y_t,z_t\geq 0.
$$

La suite est définie par les relations de récurrence suivantes.



## Partie I — 1. Écriture matricielle

### a) Expression en fonction de xt, yt, zt

$$
\begin{aligned}
x_{t+1} &= \frac12 x_t+\frac16 y_t+\frac16 z_t,\\
y_{t+1} &= \frac14 x_t+\frac23 y_t+\frac16 z_t,\\
z_{t+1} &= \frac14 x_t+\frac16 y_t+\frac23 z_t.
\end{aligned}
$$

### b) Expression matricielle

En posant

$$
A=
\begin{pmatrix}
\frac12 & \frac16 & \frac16\\
\frac14 & \frac23 & \frac16\\
\frac14 & \frac16 & \frac23
\end{pmatrix},
$$

on obtient

$$
\boxed{v_{t+1}=Av_t.}
$$

### c) Récurrence

Montrons par récurrence que, pour tout $t\in\mathbb N$,

$$
\boxed{v_t=A^tv_0.}
$$

**Initialisation :** Pour $t=0$, on a

$$
A^0v_0=I_3v_0=v_0.
$$

La propriété est donc vraie au rang $0$.

**Hérédité :** Supposons qu'à un rang $t$, on ait
$v_t=A^tv_0$. Alors, d'après la relation de récurrence,

$$
A^{t+1}v_0=A^1(A^tv_0)=A(v_t)
$$

A la question b on avait bien $$ {v_{t+1}=Av_t.} $$

La propriété est donc héréditaire.

**Conclusion :** Par récurrence,

$$
\boxed{\forall r\in\mathbb N,\quad v_t=A^tv_0.}
$$



## Partie I — 2. Propriétés de la matrice

### a) Montrer que $A$ est stochastique

Une matrice est stochastique (par colonnes) si ses coefficients sont positifs ou nuls et si la somme des coefficients de chaque colonne vaut $1$.

Les coefficients de $A$ sont positifs. Vérifions les sommes des colonnes :

$$
\begin{aligned}
\text{Colonne 1 : }&\frac12+\frac14+\frac14=1,\\
\text{Colonne 2 : }&\frac16+\frac23+\frac16=1,\\
\text{Colonne 3 : }&\frac16+\frac16+\frac23=1.
\end{aligned}
$$

Ainsi, $A$ est bien une matrice stochastique par colonnes.

### b) Conservation de la somme des coordonnées

On a

$$
v_t=
\begin{pmatrix}
x_t\\
y_t\\
z_t
\end{pmatrix}
\quad\text{et}\quad
x_t+y_t+z_t=1.
$$

Comme $v_t$ est un vecteur de proportion et $v_{t+1}=x_{t+1}+y_{t+1}+z_{t+1}$ où $x_{t+1}$, $y_{t+1}$, $z_{t+1}$ sont des sommes de proportions de $x_t$, $y_t$, $z_t$ et on a vu que les sommes de chaque coefficients sont égalesà 1 on garde toutes les proportions donc ça ne sera jamais < 0 et ça ne dépassera jamais 1

$$
\begin{aligned}
x_{t+1}+y_{t+1}+z_{t+1}
&=\left(\frac12+\frac14+\frac14\right)x_t\\
&\quad+\left(\frac16+\frac23+\frac16\right)y_t\\
&\quad+\left(\frac16+\frac16+\frac23\right)z_t\\
&=x_t+y_t+z_t=1.
\end{aligned}
$$




## Partie I — 3. Valeur propre et vecteur propre

On note

$$
\mathbf{1}=
\begin{pmatrix}
1\\
1\\
1
\end{pmatrix}.
$$

### a) Calculer $A^T\mathbf{1}$

On a

$$
A^T=
\begin{pmatrix}
\frac12 & \frac14 & \frac14\\
\frac16 & \frac23 & \frac16\\
\frac16 & \frac16 & \frac23
\end{pmatrix}.
$$

Donc

$$
\begin{aligned}
A^T\mathbf{1}
&=
\begin{pmatrix}
\frac12+\frac14+\frac14\\
\frac16+\frac23+\frac16\\
\frac16+\frac16+\frac23
\end{pmatrix}\\
&=
\begin{pmatrix}
1\\
1\\
1
\end{pmatrix}
=\mathbf{1}.
\end{aligned}
$$

Ainsi, $1$ est une valeur propre de $A^T$.

### b) En déduire que $1$ est une valeur propre de $A$

Parce que les valeurs propres sont les racines du polynome caractéristiques associées $P_A(\lambda) = \det(A-\lambda I_3)$

donc pour $A^T$ on a $P_A^T(\lambda) = \det(A^T-\lambda I_3)$ ou $\det(A^T-(\lambda I_3)^T) = \det(A-(\lambda I_3))$

Or le déterminant d'une matrice est égal au déterminant de sa transposée donc $\det(A-\lambda I_3) = \det(A^T-\lambda I_3)$

Donc ils ont le même polynome caractéristique et le même spectre donc $sp(A) = sp(A^T)$

Donc 1 est bien valeur propre de $A$

### c) Déterminer une base de $\ker(A-I_3)$

On cherche les vecteurs $v=(x,y,z)$ tels que

$$
(A-I_3)v=0.
$$

Or

$$
A-I_3=
\begin{pmatrix}
-\frac12 & \frac16 & \frac16\\
\frac14 & -\frac13 & \frac16\\
\frac14 & \frac16 & -\frac13
\end{pmatrix}.
$$

Le système associé est donc

$$
\begin{cases}
-\frac12x+\frac16y+\frac16z=0\\
\frac14x-\frac13y+\frac16z=0\\
\frac14x+\frac16y-\frac13z=0
\end{cases}
$$

En multipliant les équations respectivement par $6$, $12$ et $12$, on obtient

$$
\begin{cases}
-3x+y+z=0\\
3x-4y+2z=0\\
3x+2y-4z=0
\end{cases}
$$

$L3$ <- $L3 + L1$

$$
\begin{cases}
-3x+y+z=0\\
3x-4y+2z=0\\
3y=3z
\end{cases}
$$

Et donc : 

$$
\begin{cases}
x=\frac23y\\
y=z
\end{cases}
$$

En posant $z=3$, on obtient

$$
\begin{pmatrix}
x\\
y\\
z
\end{pmatrix}
=
\begin{pmatrix}
2\\
3\\
3
\end{pmatrix}
$$

On a 

$$
\boxed{
\ker(A-I_3)=
\operatorname{Vect}
\left\{
\begin{pmatrix}
2\\
3\\
3
\end{pmatrix}
\right\}.
}
$$

```python
def is_stochastic(A, orientation="column", atol=1e-12):

	# Vérifie si la matrice est carrée à partir du compte de ses éléments en ligne et colonne
	if A.shape[0] != A.shape[1]:
		print(False)

	# Vérifie que les coefficients soient positifs ou nuls
	if np.any(A < -atol):
		return (False)

	# Vérifie que la somme des éléments en colonne soit 1 à la tolérance près
	if orientation == "column":
		return np.allclose(A.sum(axis=0), 1, atol=atol)

    # Même opération sur les lignes 
	if orientation == "row":
		return np.allclose(A.sum(axis=1), 1, atol=atol)
    	
	return False
```

La transposée de A et la matrice B renvoient un test "False" parce qu'il ne s'agit pas de matrices stochastiques en colonnes.

```python
def simulate_markov(A, x0, T):

	# 3 vérifications préalables : A est carrée, le vecteur x0 est de même dimension que A, T ne permet pas le calcul 
	if A.shape[0] != A.shape[1] or len(x0) != A.shape[0] or T < 0:
        	raise ValueError("Erreur dans la compatibilité des dimensions")

	# Prise en compte de l'état initial
	etats = [x0]
	x = x0.copy()

	# Répétition T fois du produit Ax et enregistrement du nouveau résultat
	for t in range(T):
		x = A @ x
		etats.append(x.copy())

	# Résultat intégré en tableau NumPy
	return np.array(etats)
```

```python
import numpy as np

def eigen_data(A):

    valeurs_propres, vecteurs_propres = np.linalg.eig(A)

    # On veut trier par ordre décroissant de module des valurs propres donc on prend dans l'ordre avec argsort et on trie par - valeur absolue!
    indices = np.argsort(-np.abs(valeurs_propres))

    # On récupère les indices dans les arrays pour ensuite faire le np.real_if_close de correction d'érreurs d'arrondi
    valeurs_propres = valeurs_propres[indices]                 
    vecteurs_propres = vecteurs_propres[:, indices]

    valeurs_propres = np.real_if_close(valeurs_propres)
    vecteurs_propres = np.real_if_close(vecteurs_propres)

    return valeurs_propres, vecteurs_propres
```

```python
import numpy

def diagonalize(A, atol=1e-12):

    valeurs, vecteurs = eigen_data(A)

    # On construit les matrices P et D pour appliquer la formule : P est l'ensemble des vecteurs propres en matrice et D la matrice diagonale des valeurs propres!
    P = vecteurs
    D = np.diag(valeurs)

    # On vérifie si P est bien inversible?
    if np.linalg.matrix_rank(P, tol=atol) < A.shape[0]:
        raise ValueError("Matrice A non diagonalisable!")

    return P, D
```

6)     Méthode 1 :

```python
def power_iterative(A, n):

    # On créée une matrice identité de même dimension que A
    matrice_id = np.eye(A.shape[0])

    # On calcule n fois A x A, si n = 0 on retourne matrice identité, si n=1 -> A x id !
    for i in range(n):
        matrice_id = matrice_id @ A

    return matrice_id
```

6)     Méthode 2 :

```python
def power_diagonalized(P, D, n):

    # On calcule Dn directement sur la diagonale D n-fois
    Dn = np.diag(np.diag(D)**n)

    # On calcule An, avec les termes précédents et on inverse P qu'on a bien vérifié inversible!
    An = P @ Dn @ np.linalg.inv(P)

    return An
```

6)     Méthode 3 :

```python
def power_numpy(A, n):

    # Fonction qui se trouve directement dans le package numpy
    return np.linalg.matrix_power(A, n)
```

6)    Comparaison des méthodes

```python
import numpy as np
import matplotlib.pyplot as plt

# On fixe des paramètres de base, (à changer plus tard potentiellement)
Nmax = 50
seuil_erreur = 1e-6

# On utilise notre fonction pour diagonaliser A
P, D = diagonalize(A)

# On a trois 3 ecarts pour les 3 fonctions donc on 3 listes contenant les erreurs
erreurs_iterative_power = []
erreurs_diagonalized_power  = []
erreurs_diagonalized_numpy = []

# On calcule les erreurs pour n jusqu'à Nmax : on fait nos trois méthodes, puis on met les erreurs dans les listes créées préalablement avec la norme de Frobenius!
for n in range(Nmax + 1):

    A_iterative = power_iterative(A, n)
    A_diagonalized = power_diagonalized(P, D, n)
    A_numpy = power_numpy(A, n)

    erreurs_iterative_power.append(np.linalg.norm(A_iterative - A_diagonalized, ord='fro'))
    erreurs_diagonalized_power.append(np.linalg.norm(A_iterative - A_numpy, ord='fro'))
    erreurs_diagonalized_numpy.append(np.linalg.norm(A_diagonalized - A_numpy, ord='fro'))
```

6) Première valeur n pour laquelle c'est inférieur à 10**-6

```python
def seuil_erreur(erreurs_liste, seuil=1e-6):

    erreurs_liste = np.array(erreurs_liste)

    # On trouve l'indice dans la liste a partir duquel l'erreur est inférieur au seuil
    liste_indices = np.where(erreurs_liste < seuil)[0]

    if len(liste_indices) == 0:
        print("Le seuil n'est pas atteint")
    else:
        premier_n_seuil = liste_indices[0]

        # On vérifie si tous les autres éléments de la liste vérifient le seuil!
        if np.all(erreurs_liste[premier_n_seuil:] < seuil):
            print("L'erreur reste inférieure au seuil")
        else:
            print("L'erreur ne reste pas inférieure au seuil")
    
    return premier_n_seuil
```

```python
seuil_erreur(erreurs_iterative_power)
seuil_erreur(erreurs_diagonalized_power)
seuil_erreur(erreurs_diagonalized_numpy)
```
